## Getting Postcode District

```python
# Function to get postcode district from postcode
def Postcode_District(Postcode):
# Normal scenario
    if " " in Postcode:
        Split = Postcode.split()
        District = Split[0].upper()
# Bad formatting
    else:
        Outward = Postcode[:-1].strip("_").strip("-").strip("+")
        District = Outward.upper()
    return (District)
```

```python
# Add postcode district and postcode sector into dataframe
def Add_Postcode_District(DF_orig):
    DF = DF_orig.copy()
    Districts = []
# Find postcode district for each postcode sector in dataframe
    for i in range(len(DF)):
        Districts.append(Postcode_District(DF['postcodeSector'][i]))
# Add new column to dataframe
    DF['postcodeDistrict'] = Districts
    return (DF)
# Apply function
DF_updated = Add_Postcode_District(DF_orig)
DF_updated
```
